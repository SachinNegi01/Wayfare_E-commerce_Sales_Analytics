```python
import pandas as pd
import matplotlib.pyplot as plt
from pathlib import Path
```

```python
data_dir = Path("datasets")
```

```python
customers = pd.read_csv(data_dir / "customers.csv")
order_items = pd.read_csv(data_dir / "order_items.csv")
orders = pd.read_csv(data_dir / "orders.csv")
products = pd.read_csv(data_dir / "products.csv")
```

```python
print("customers:", customers.shape)
print("order_items:", order_items.shape)
print("orders:", orders.shape)
print("products:", products.shape)
```

```python
for name, df in {
    "customers": customers,
    "order_items": order_items,
    "orders": orders,
    "products": products
}.items():
    print(f"\n--- {name} ---")
    print("Columns:", df.columns.tolist())
    print("\ndata types and non_missing counts:")
    df.info()
    print("\nFirst 5 rows:")
    display(df.head())
```

```python
for name, df in {
    "customers": customers,
    "order_items": order_items,
    "orders": orders,
    "products": products
}.items():
    print(f"\n--- {name}: missing values ---")
    print(df.isna().sum())
    print(f"\n--- {name}: exact duplicate rows ---")
    print(df.duplicated().sum())
```

```python
print("Repeated order IDs:", orders["order_id"].duplicated().sum())
print("\nRepeated order_id rows:")
display(orders[orders.duplicated(subset="order_id", keep=False)]
        .sort_values("order_id")
        .head(15))

print("\nMissing unit_price rows:")
display(order_items[order_items["unit_price"].isna()].head(10))

print("\nOrder status values:")
print(orders["status"].value_counts(dropna=False))

print("\nDiscount code examples:")
print(orders["discount_code"].value_counts(dropna=False).head(15))
```

```python
duplicate_orders = orders[
    orders.duplicated(subset="order_id", keep=False)
].sort_values("order_id")

print("Rows with repeated order IDs:", len(duplicate_orders))
print(
    "Distinct full rows among repeated IDs:",
    len(duplicate_orders.drop_duplicates())
)

print("\nRepeated IDs with differing order details:")
display(
    duplicate_orders.groupby("order_id")
    .nunique()
    .drop(columns="order_id", errors="ignore")
    .query("customer_id > 1 or order_date > 1 or status > 1 "
           "or payment_method > 1 or discount_code > 1 or shipping_fee > 1")
)

print("\nMissing-price rows with matching product list prices:")
missing_prices = order_items[order_items["unit_price"].isna()].merge(
    products[["product_id", "list_price"]],
    on="product_id",
    how="left"
)
display(missing_prices.head(10))
```
